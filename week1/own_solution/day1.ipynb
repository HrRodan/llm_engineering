```python

from dotenv import load_dotenv
from openai import OpenAI
import os
import sys
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..','..')))
from tools import get_gemini_client

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

system_prompt_bob = "You are cool dude named Bob, intelligent and helpful. Always answer in a friendly way. Begin every answer with 'Bob:'."
user_prompt_bob = "Your have a conversion with Alice. Keep the conversation going and ask questions. The complete history of the conversation is given below (empty if start of conversation):\n"

system_prompt_alice = "You are Alice, humorful and a bit lazy. Always answer in a friendly way. Begin every answer with 'Alice:'."
user_prompt_alice = "Your have a conversion with Bob. Keep the conversation going and ask questions. The complete history of the conversation is given below (empty if start of conversation):\n"

def generate_message(system_prompt, user_prompt, conversation_history = []):
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt + "\n" + "\n".join(conversation_history)}
    ]
    return messages

conversation_history = []


```

```python
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"

load_dotenv(override=True)

google_api_key = os.getenv("GOOGLE_API_KEY")

gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)

messages = generate_message(system_prompt_bob, user_prompt_bob)

response = gemini.chat.completions.create(model="gemini-3-pro-preview", messages=messages)
response.choices[0].message.content

conversation_history.append(response.choices[0].message.content)
```

```python
for _ in range(2):
    messages_alice = generate_message(system_prompt_alice, user_prompt_alice, conversation_history)
    response_alice = gemini.chat.completions.create(
        model="gemini-3-pro-preview",
        messages=messages_alice,
    )
    print(response_alice.choices[0].message.content)
    conversation_history.append(response_alice.choices[0].message.content)

    messages_bob = generate_message(system_prompt_bob, user_prompt_bob, conversation_history)
    response_bob =  gemini.chat.completions.create(
        model="gemini-3-pro-preview",
        messages=messages_bob,
    )
    conversation_history.append(response_bob.choices[0].message.content)
    print(response_bob.choices[0].message.content)

print('\n'.join(conversation_history))
```
