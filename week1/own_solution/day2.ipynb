```python
from openai import OpenAI
import os
import sys
from openai import OpenAI
from  scraper import fetch_website_contents
from tools import get_gemini_client

sytem_prompt_page_summarzier = "You summarzize Web Pages. You are given the complete webpage, ignore all navigation and similar elements not relevant for the content"
user_prompt_page_summarzier = "Summarize the following web page: "

def generate_message(system_promt, user_prompt, content):
    return [
        {"role": "system", "content": system_promt},
        {"role": "user", "content": user_prompt + content}
    ]

URL = "https://www.bbc.com/news/articles/c2k4z9nwz4yo"

webpage_content = fetch_website_contents(URL)
```

```python

OLLAMA_BASE_URL = "http://localhost:11434/v1"
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='ollama')

respond = ollama.chat.completions.create( # pyrefly: ignore
    model="deepseek-r1:1.5b",
    messages=generate_message(sytem_prompt_page_summarzier, user_prompt_page_summarzier, webpage_content)) 

print(respond.choices[0].message.content)
```
