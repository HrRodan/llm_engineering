```python
# Constants

BASE_MODEL = "meta-llama/Llama-3.2-3B"
PROJECT_NAME = "price"
HF_USER = "Rodan009" # your HF name here!

LITE_MODE = True

DATA_USER = "ed-donner" if not LITE_MODE else "Rodan009"
DATASET_NAME = f"{DATA_USER}/items_prompts_lite" if LITE_MODE else f"{DATA_USER}/items_prompts_full"

RUN_NAME =  f"{datetime.now():%Y-%m-%d_%H.%M.%S}"
if LITE_MODE:
  RUN_NAME += "-lite"
PROJECT_RUN_NAME = f"{PROJECT_NAME}-{RUN_NAME}"
HUB_MODEL_NAME = f"{HF_USER}/{PROJECT_RUN_NAME}"

# Hyper-parameters - overall

EPOCHS = 1 if LITE_MODE else 3
BATCH_SIZE = 46 if LITE_MODE else 256
MAX_SEQUENCE_LENGTH = 143 # from Day2
GRADIENT_ACCUMULATION_STEPS = 1

# Hyper-parameters - QLoRA

QUANT_4_BIT = True
LORA_R = 32 if LITE_MODE else 256
LORA_ALPHA = LORA_R * 2
ATTENTION_LAYERS = ["q_proj", "v_proj", "k_proj", "o_proj"]
MLP_LAYERS = ["gate_proj", "up_proj", "down_proj"]
TARGET_MODULES = ATTENTION_LAYERS if LITE_MODE else ATTENTION_LAYERS + MLP_LAYERS
LORA_DROPOUT = 0.1

# Hyper-parameters - training

LEARNING_RATE = 1e-4
WARMUP_RATIO = 0.01
LR_SCHEDULER_TYPE = 'cosine'
WEIGHT_DECAY = 0.001
OPTIMIZER = "paged_adamw_32bit"

capability = torch.cuda.get_device_capability()
use_bf16 = capability[0] >= 8

# Tracking

VAL_SIZE = 500 if LITE_MODE else 1000
LOG_STEPS = 5 if LITE_MODE else 10
SAVE_STEPS = 100 if LITE_MODE else 200
LOG_TO_WANDB = True
```

## Now load the Tokenizer and Model

The model is "quantized" - we are reducing the precision to 4 bits.

```python
# pick the right quantization

if QUANT_4_BIT:
  quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16 if use_bf16 else torch.float16,
    bnb_4bit_quant_type="nf4"
  )
else:
  quant_config = BitsAndBytesConfig(
    load_in_8bit=True,
    bnb_8bit_compute_dtype=torch.bfloat16 if use_bf16 else torch.float16,
  )
```

```python
# Load the Tokenizer and the Model

tokenizer = AutoTokenizer.from_pretrained(BASE_MODEL, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL,
    quantization_config=quant_config,
    device_map="auto",
)
base_model.generation_config.pad_token_id = tokenizer.pad_token_id

print(f"Memory footprint: {base_model.get_memory_footprint() / 1e6:.1f} MB")
```

# AND NOW

## We set up the configuration for Training

We need to create 2 objects:

A LoraConfig object with our hyperparameters for LoRA

An SFTConfig with our overall Training parameters

```python
# LoRA Parameters

lora_parameters = LoraConfig(
    lora_alpha=LORA_ALPHA,
    lora_dropout=LORA_DROPOUT,
    r=LORA_R,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=TARGET_MODULES,
)
```

```python
train_parameters = SFTConfig(
    # The output directory where the model predictions and checkpoints will be written
    output_dir=PROJECT_RUN_NAME,

    # Total number of training epochs to perform (passes through the entire dataset)
    num_train_epochs=EPOCHS,

    # The batch size per GPU/TPU core/CPU for training
    per_device_train_batch_size=BATCH_SIZE,

    # The batch size per GPU/TPU core/CPU for evaluation
    per_device_eval_batch_size=1,

    # Number of update steps to accumulate the gradients for, before performing a backward/update pass
    # (Effective batch size = per_device_train_batch_size * gradient_accumulation_steps)
    gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,

    # The optimizer to use (e.g., "paged_adamw_8bit", "adamw_torch")
    optim=OPTIMIZER,

    # Number of updates steps before saving a checkpoint
    save_steps=SAVE_STEPS,

    # Limit the total amount of checkpoints. Deletes the older checkpoints in output_dir.
    save_total_limit=10,

    # Number of update steps between two logs
    logging_steps=LOG_STEPS,

    # The initial learning rate for the optimizer (e.g., AdamW)
    learning_rate=LEARNING_RATE,

    # Weight decay to apply to all layers except bias/LayerNorm weights (regularization technique)
    weight_decay=0.001,

    # Whether to use fp16 16-bit (mixed) precision training (requires NVIDIA GPU)
    fp16=not use_bf16,

    # Whether to use bf16 16-bit (mixed) precision training (requires Ampere or newer NVIDIA GPU)
    bf16=use_bf16,

    # Maximum gradient norm (for gradient clipping) to prevent exploding gradients
    max_grad_norm=0.3,

    # If > 0: set total number of training steps to perform. Overrides num_train_epochs.
    # Set to -1 to use num_train_epochs instead.
    max_steps=-1,

    # Ratio of total training steps used for a linear warmup from 0 to learning_rate
    warmup_ratio=WARMUP_RATIO,

    # Whether or not to group samples of roughly the same length together when batching
    # (minimizes padding and speeds up training)
    group_by_length=True,

    # The scheduler type to use (e.g., "cosine", "linear", "constant")
    lr_scheduler_type=LR_SCHEDULER_TYPE,

    # The list of integrations to report the results and logs to (e.g., "wandb", "tensorboard")
    report_to="wandb" if LOG_TO_WANDB else None,

    # The name of the run (used for logging platforms like Weights & Biases)
    run_name=RUN_NAME,

    # The maximum permitted length for sequences in the batch
    max_length=MAX_SEQUENCE_LENGTH,

    # The checkpoint save strategy to adopt during training ("steps", "epoch", or "no")
    save_strategy="steps",

    # Defines the scope of what is pushed to the Hub and when ("every_save", "end", etc.)
    hub_strategy="every_save",

    # Whether or not to push the model to the Hugging Face Hub every time the model is saved
    push_to_hub=True,

    # The name of the repository to keep in sync with the local output_dir
    hub_model_id=HUB_MODEL_NAME,

    # Whether the model repository is private or not
    hub_private_repo=True,

    # The evaluation strategy to adopt during training ("steps", "epoch", or "no")
    eval_strategy="steps",

    # Number of update steps between two evaluations if eval_strategy="steps"
    eval_steps=SAVE_STEPS
)
```

# AND NOW - create the trainer

```python
fine_tuning = SFTTrainer(
    model=base_model,
    train_dataset=train,
    eval_dataset=val,
    peft_config=lora_parameters,
    args=train_parameters
)
```
