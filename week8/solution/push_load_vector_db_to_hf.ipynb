```python
import os
from huggingface_hub import HfApi, login

# Ensure you are logged in. If not, uncomment the next line and enter your token:
hf_token = os.environ['HF_TOKEN']
login(token=hf_token, add_to_git_credential=False)

# Configuration
username = "Rodan009"
folder_path = "products_vectorstore"
dataset_name = "items_full_vectorstore"
repo_id = f"{username}/{dataset_name}"

try:
    print(f"Target Repository: {repo_id}")
    api = HfApi()

    # Create the repository (if it doesn't exist)
    api.create_repo(repo_id=repo_id, repo_type="dataset", exist_ok=True)

    # Upload the folder
    print(f"Uploading '{folder_path}' to {repo_id}...")
    api.upload_folder(
        folder_path=folder_path,
        repo_id=repo_id,
        repo_type="dataset",
        path_in_repo="."  # Upload contents to the root of the dataset
    )
    print("Upload complete!")

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from huggingface_hub import snapshot_download
import shutil
import os

# Configuration
username = "Rodan009"
dataset_name = "items_full_vectorstore"
repo_id = f"{username}/{dataset_name}"
target_dir = "products_vectorstore"

try:
    print(f"Downloading from {repo_id}...")
    
    # Download the dataset from Hugging Face (stored in cache)
    cache_path = snapshot_download(repo_id=repo_id, repo_type="dataset")
    print(f"Snapshot downloaded to cache: {cache_path}")

    # Move/Copy to a local usable directory
    if os.path.exists(target_dir):
        print(f"Removing existing directory {target_dir}...")
        shutil.rmtree(target_dir)
    
    print(f"Copying to local path: {target_dir}...")
    shutil.copytree(cache_path, target_dir)
    
    print(f"Successfully loaded vector store to '{target_dir}'")

except Exception as e:
    print(f"An error occurred: {e}")
```
